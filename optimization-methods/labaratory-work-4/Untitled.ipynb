```python
def bellman_ford(vertices, edges, start_node):
    # 1. Инициализация расстояний
    distances = {v: float('inf') for v in vertices}
    distances[start_node] = 0
    
    # 2. Основной цикл релаксации (V - 1 раз)
    for i in range(len(vertices) - 1):
        for u, v, weight in edges:
            if distances[u] != float('inf') and distances[u] + weight < distances[v]:
                distances[v] = distances[u] + weight
                
    # 3. Проверка на наличие отрицательных циклов
    for u, v, weight in edges:
        if distances[u] != float('inf') and distances[u] + weight < distances[v]:
            print("Внимание: Граф содержит цикл отрицательного веса!")
            return None
            
    return distances

# Данные нашего примера
nodes = ['A', 'B', 'C', 'D']
graph_edges = [
    ('A', 'B', 5),
    ('A', 'C', 2),
    ('B', 'D', 1),
    ('C', 'B', -2),
    ('C', 'D', 6)
]

# Запуск
result = bellman_ford(nodes, graph_edges, 'A')

if result:
    print("Кратчайшие расстояния от вершины A:")
    for node, dist in result.items():
        print(f"До {node}: {dist}")
```
