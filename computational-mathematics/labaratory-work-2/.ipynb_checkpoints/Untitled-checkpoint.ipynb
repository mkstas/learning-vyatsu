## Задание 1. Решить систему линейных уравнений методом Гаусса

Дана система линейныйх уравнений 4-го порядка, $\varepsilon=0.001$:

\begin{cases}
1.32x_1 - 0.83x_2 - 0.44x_3 + 0.62x_4 = 0.68 \\
0.83x_1 + 0.42x_2 - 0.56x_3 + 0.77x_4 = 1.24 \\
0.58x_1 - 0.37x_2 + 1.24x_3 - 0.62x_4 = 0.87 \\
0.35x_1 + 0.66x_2 - 1.38x_3 + 0.93x_4 = 1.08
\end{cases}

Зададим исходные данные матрицы:

```python
A = np.array([
    [1.32, -0.83, -0.44,  0.62],
    [0.83,  0.42, -0.56,  0.77],
    [0.58, -0.37,  1.24, -0.62],
    [0.35,  0.66, -1.38,  0.93]
])

b = np.array([0.68, 1.24, 0.87, 1.08])
```

Напишем функцию для решения СЛАУ с помощью метода Гаусса

```python
def gauss(A, b, use_pivoting=True, use_steps=False):
    n = len(b)
    matrix = np.column_stack((A.astype(float), b.astype(float)))

    for i in range(n):
        if use_pivoting:
            max_row = np.argmax(np.abs(matrix[i:, i])) + i
            matrix[[i, max_row]] = matrix[[max_row, i]]

        for j in range(i + 1, n):
            factor = matrix[j, i] / matrix[i, i]
            matrix[j, i:] -= factor * matrix[i, i:]

        if use_steps:
            print(matrix, "\n")

    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (matrix[i, n] - np.dot(matrix[i, i+1:n], x[i+1:n])) / matrix[i, i]
    return x
```

Решим СЛАУ с выбором ведущего элемента

```python
x = gauss(A, b, use_pivoting=True, use_steps=True)
r = residual(A, x, b)
print(f"Решение СЛАУ:\n{x}\n")
print(f"Вектор невязки:\n{r}")
```

Решим СЛАУ без выбора ведущего элемента

```python
x = gauss(A, b, use_pivoting=False, use_steps=True)
r = residual(A, x, b)
print(f"Решение СЛАУ:\n{x}\n")
print(f"Вектор невязки:\n{r}")
```

Результаты решения с выбором и без выбора главного элемента не отличаются.

Изменим исходные коэффиценты

```python
A = np.array([
    [0.000000000001, -0.83, -0.44,  0.62],
    [0.83,  0.42, -0.56,  0.77],
    [0.58, -0.37,  1.24, -0.62],
    [0.35,  0.66, -1.38,  0.93]
])

b = np.array([0.68, 1.24, 0.87, 1.08])

x = gauss(A, b, use_pivoting=True, use_steps=False)
r = residual(A, x, b)
print(f"Решение СЛАУ с выбором ведущего элемента:\n{x}\n")
print(f"Вектор невязки:\n{r}\n")

x = gauss(A, b, use_pivoting=False, use_steps=False)
r = residual(A, x, b)
print(f"Решение СЛАУ без выбора ведущего элемента:\n{x}\n")
print(f"Вектор невязки:\n{r}")
```

Ипользуем в качестве примера матрицу Гильберта размерностью 4x4

```python
H = la.hilbert(4)
x = np.array([1.0, 2.0, 3.0, 4.0])
b = gauss(H, x, use_pivoting=True, use_steps=False)

print(f"Число обусловленности матрицы Гильберта: {np.linalg.cond(H):.4f}", "\n")
print(f"Решение системы Hx = b:\n{b}\n")
print(f"Число обусловленности матрицы b: {np.linalg.cond([b]):.4f}", "\n")
```

## Задание 2. Решить систему линейных уравнений методом простых итераций

Дана система линейныйх уравнений 4-го порядка, $\varepsilon=0.0001$:

\begin{cases}
x_1 = 0.13x_1 + 0.23x_2 - 0.44x_3 - 0.05x_4 + 2.13 \\
x_2 = 0.24x_1 - 0.31x_3 + 0.15x_4 + 0.18 \\
x_3 = 0.06x_1 + 0.15x_2 - 0.23x_4 + 1.44 \\
x_4 = 0.72x_1 - 0.08x_2 - 0.05x_3 + 2.42
\end{cases}

Зададим исходные данные матрицы:

```python
B = np.array([
    [0.13, 0.23, -0.44, -0.05],
    [0.24, 0.00, -0.31, 0.15],
    [0.06, 0.15, 0.00, -0.23],
    [0.72, -0.08, -0.05, 0.00]
])

c = np.array([2.13, 0.18, 1.44, 2.42])
```

Проверим условия сходимости метода порстых итераций

```python
row_norm = np.max(np.sum(np.abs(B), axis=1))
col_norm = np.max(np.sum(np.abs(B), axis=0))
euclid_norm = np.sqrt(np.sum(B ** 2))

if row_norm <= 1 or col_norm <= 1 or euclid_norm <= 1:
    print("Итерационный метод сходится, т.к. выполнено достаточное условие норм")
else:
    print("Итерационный метод не сходится, т.к. не выполнено достаточное условие норм")
```

Напишем функцию для решения СЛАУ с помощью метода простых итераций

```python
def simple_iteration(B, c, eps=1e-5, max_iter=100):
    x = c.copy()
    for i in range(max_iter):
        x_new = B @ x + c
        diff_vec = np.abs(x_new - x)

        if np.all(diff_vec <= eps):
            return x_new, i + 1

        x = x_new

    return x, max_iter
```

```python
x, i = simple_iteration(B, c)
r = A @ x + b - x
print(f"Решение СЛАУ\n{x}\n")
print(f"Количество итераций: {i}\n")
print(f"Вектор невязки:\n{r}\n")
```

## Задание 3. Решить систему линейных уравнений методом обратной матрицы

Дана система линейныйх уравнений 3-го порядка, $\varepsilon=0.001$:

\begin{cases}
-2x_1 + 3x_2 + 2x_3 = 4 \\
4x_1 + 3x_2 - 9x_3 = -4 \\
8x_1 + 3x_2 - 5x_3 = 2
\end{cases}

Зададим исходные данные матрицы:

```python
A = np.array([
    [-2, 3, 2],
    [4, 3, -9],
    [8, 3, -5]
])

b = np.array([4, -4, 2])
```

Напишем функцию для нахождения обратной матрицы:

```python
def gauss_inverse(A):
    n = len(A)
    matrix = np.hstack((A.astype(float), np.eye(n)))

    for i in range(n):
        max_row = np.argmax(np.abs(matrix[i:, i])) + i
        matrix[[i, max_row]] = matrix[[max_row, i]]
        matrix[i] = matrix[i] / matrix[i, i]

        for j in range(n):
            if i != j:
                factor = matrix[j, i]
                matrix[j] -= factor * matrix[i]

    return matrix[:, n:]
```

Напишем функцию для решения методом обратной матрицы

```python
def inverse(A, b):
    A_inv = gauss_inverse(A)
    x = A_inv @ b
    return x, A_inv
```

```python
x, A_inv = inverse(A, b)
r = residual(A, x, b)
print(f"Решение СЛАУ:\n{x}\n")
print(f"Вектор невязки:\n{r}\n")
```

## Задание 4. Решить систему нелинейных уравнений методом Ньютона

Дана система нелинейныйх уравнений 2-го порядка, $\varepsilon=0.001$:

\begin{cases}
5x + tg(x * y) = 0 \\
(y^2 - 1.5)^2 + ln(x) = 0
\end{cases}

```python
import numpy as np
import matplotlib.pyplot as plt

x_range = np.linspace(0.001, 1.5, 1000) 
y_range = np.linspace(-3.0, 3.0, 1000)
X, Y = np.meshgrid(x_range, y_range)

F1 = 5 * X + np.tan(X * Y)
F2 = (Y ** 2 - 1.5) ** 2 + np.log(X)
F1_masked = np.ma.masked_where(np.abs(F1) > 20, F1)

plt.figure(figsize=(10, 6))
plt.axhline(0, color='black', linewidth=1, linestyle='-')
plt.axvline(0, color='black', linewidth=1, linestyle='-')
plt.grid(True, linestyle='-', alpha=0.5)
plt.xlim(0.001, 3)
plt.ylim(-4.0, 4.0)

plt.contour(X, Y, F1_masked, levels=[0], colors="blue", linewidths=2)
plt.contour(X, Y, F2, levels=[0], colors="red", linewidths=2)

plt.show()
```

Зададим исходные данные матрицы:

```python
def F(x):
    return np.array([
        5 * x[0] + np.tan(x[0] * x[1]),
        (x[1] ** 2 - 1.5) ** 2 + np.log(x[0])
    ])

def J(x):
    tan_val = np.tan(x[0] * x[1])

    return np.array([
        [5 + x[1] * (1 + tan_val ** 2), x[0] * (1 + tan_val ** 2)],
        [1 / x[0], 4 * x[1] * (x[1]**2 - 1.5)]
    ])
```

Напишем функцию для решения методом Ньютона

```python
def newton(eps=1e-4, max_iter=100):
    x = np.array([0.001, 1.5])

    for i in range(max_iter):
        delta = np.linalg.solve(J(x), -F(x))
        x_new = x + delta

        step_factor = 1.0
        while x_new[0] <= 0 and step_factor > eps:
            step_factor *= 0.5
            x_new = x + delta * step_factor

        if np.max(np.abs(delta)) <= eps:
            return x_new, i + 1

        x = x_new

    return x, max_iter
```

```python
x, i = newton()
r = F(x)
print(f"Решение:\n{x}\n")
print(f"Количество итераций: {i}\n")
print(f"Вектор невязки:\n{r}\n")
```

## LU-разложение

Необходимо выполнить LU-разложение $(PA=LU)$ матрицы из задания 1 с помощью библиотеки linalg.<br>
Решить $Ax = b$ для 100 случайных векторов.<br>
$P$ - матрица для предотвращения деления на ноль, уменьшить количество ошибок округления и сохранения порядка перестановки строк.<br>
$Ax = b$ &rArr; $LUx = Pb$:
- $Ly = Pb$ - прямой ход
- $Ux = y$ - обратный ход

```python
A = np.array([
    [1.32, -0.83, -0.44,  0.62],
    [0.83,  0.42, -0.56,  0.77],
    [0.58, -0.37,  1.24, -0.62],
    [0.35,  0.66, -1.38,  0.93]
])

b = np.array([0.68, 1.24, 0.87, 1.08])
```

```python
def lu(P, L, U, b):
    n = len(b)
    Pb = P @ b

    y = np.zeros(n)
    for i in range(n):
        y[i] = Pb[i] - sum(L[i, j] * y[j] for j in range(i))

    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (y[i] - sum(U[i, j] * x[j] for j in range(i + 1, n))) / U[i, i]

    return x
```

Решим СЛАУ из первого задания:

```python
P, L, U = la.lu(A)
x = lu(P, L, U, b)

print(x)
```

Замерим скорость выполения для 100 случайных векторов:

```python
random_b = np.random.rand(100, 4)

start = time.time()
for i in range(100):
    x = lu(P, L, U, random_b[i])
stop = time.time() - start

print(f"Время работы для 100 случайных векторов: {stop:.3f} сек.")
```
