## Задание 1. Решить систему линейных уравнений методом Гаусса

Дана система линейныйх уравнений 4-го порядка, $\varepsilon=0.001$:

\begin{cases}
1.32x_1 - 0.83x_2 - 0.44x_3 + 0.62x_4 = 0.68 \\
0.83x_1 + 0.42x_2 - 0.56x_3 + 0.77x_4 = 1.24 \\
0.58x_1 - 0.37x_2 + 1.24x_3 - 0.62x_4 = 0.87 \\
0.35x_1 + 0.66x_2 - 1.38x_3 + 0.93x_4 = 1.08
\end{cases}

Зададим исходные матрицы:

```python
import numpy as np

A = np.array([
    [1.32, -0.83, -0.44,  0.62],
    [0.83,  0.42, -0.56,  0.77],
    [0.58, -0.37,  1.24, -0.62],
    [0.35,  0.66, -1.38,  0.93]
])

B = np.array([0.68, 1.24, 0.87, 1.08])
```

```python
import numpy as np

# Настройка вывода
np.set_printoptions(precision=4, suppress=True)

def solve_gauss(A_orig, b_orig, use_pivoting=True):
    n = len(b_orig)
    # Создаем копию, чтобы не изменять исходные данные
    A = A_orig.astype(float)
    b = b_orig.astype(float)
    matrix = np.column_stack((A, b))
    
    mode_str = "С ВЫБОРОМ главного элемента" if use_pivoting else "БЕЗ ВЫБОРА главного элемента"
    print(f"\n{'='*50}\nРЕЖИМ: {mode_str}\n{'='*50}")

    # Прямой ход
    for i in range(n):
        if use_pivoting:
            # Поиск максимального по модулю элемента в текущем столбце
            max_row = np.argmax(np.abs(matrix[i:, i])) + i
            if i != max_row:
                matrix[i], matrix[max_row] = matrix[max_row].copy(), matrix[i].copy()
                print(f"Шаг {i+1}: Меняем строку {i} со строкой {max_row} (max эл-т: {matrix[i, i]:.4f})")
        else:
            # Проверка на ноль на диагонали (базовая защита)
            if abs(matrix[i, i]) < 1e-12:
                print(f"Внимание: Нулевой элемент на диагонали на шаге {i+1}. Без пивотирования решение может быть неверным.")

        # Исключение элементов под диагональю
        for k in range(i + 1, n):
            factor = matrix[k, i] / matrix[i, i]
            matrix[k, i:] -= factor * matrix[i, i:]
        
        print(f"Матрица после шага {i+1}:")
        print(matrix)
        print("-" * 30)

    # Обратный ход
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (matrix[i, n] - np.dot(matrix[i, i+1:n], x[i+1:n])) / matrix[i][i]
    
    # Расчет невязки
    res = b_orig - np.dot(A_orig, x)
    
    return x, res

# --- Исходные данные ---
A = np.array([
    [1.32, -0.83, -0.44,  0.62],
    [0.83,  0.42, -0.56,  0.77],
    [0.58, -0.37,  1.24, -0.62],
    [0.35,  0.66, -1.38,  0.93]
])

b = np.array([0.68, 1.24, 0.87, 1.08])

# --- Решение двумя способами ---

# 1. Без выбора главного элемента
sol_no_piv, res_no_piv = solve_gauss(A, b, use_pivoting=False)

# 2. С выбором главного элемента
sol_piv, res_piv = solve_gauss(A, b, use_pivoting=True)

# --- Итоговое сравнение ---
print("\n" + "#"*50)
print("СРАВНЕНИЕ РЕЗУЛЬТАТОВ")
print("#"*50)
print(f"{'Переменная':<12} | {'Без выбора':<15} | {'С выбором':<15}")
print("-" * 46)
for i in range(len(sol_piv)):
    print(f"x{i+1:<10} | {sol_no_piv[i]:<15.6f} | {sol_piv[i]:<15.6f}")

print("\nМАКСИМАЛЬНАЯ НЕВЯЗКА:")
print(f"Без выбора: {np.max(np.abs(res_no_piv)):.2e}")
print(f"С выбором:  {np.max(np.abs(res_piv)):.2e}")
```
